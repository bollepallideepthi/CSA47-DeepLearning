```python
import numpy as np

X = np.array([1., 2., 3., 4.])
y = 2 * X

def train(mode):
    w = 0.0

    for epoch in range(30):
        ids = np.arange(len(X))

        if mode == "SGD":
            np.random.shuffle(ids)
            batches = [[i] for i in ids]
        elif mode == "Mini-batch":
            np.random.shuffle(ids)
            batches = [ids[i:i+2] for i in range(0, 4, 2)]
        else:
            batches = [ids]

        for batch in batches:
            grad = np.mean(2 * X[batch] * (w * X[batch] - y[batch]))
            w -= 0.01 * grad

    return w

for mode in ["Batch", "SGD", "Mini-batch"]:
    print(mode, train(mode))
```

```python
import numpy as np

Q = np.zeros((3, 2))

for episode in range(1000):
    state = np.random.randint(3)

    for step in range(10):
        if np.random.rand() > 0.1:
            action = np.argmax(Q[state])
        else:
            action = np.random.randint(2)

        reward = (state + 1) if action == 1 else -2 * (state + 1)

        next_state = max(0, min(2, state + np.random.choice([-1, 0, 1])))

        Q[state, action] += 0.1 * (
            reward + 0.9 * np.max(Q[next_state]) - Q[state, action]
        )

        state = next_state

print("Q Table:\n", Q)
print("Best Actions:", np.argmax(Q, axis=1))
```

```python
import numpy as np

X = np.array([[1.0, 0.5]])
target = np.array([[1.0]])

W1 = np.array([[0.1, 0.2], [0.3, 0.4]])
W2 = np.array([[0.5], [0.6]])

sigmoid = lambda x: 1 / (1 + np.exp(-x))

h = sigmoid(X @ W1)
o = sigmoid(h @ W2)

d2 = (o - target) * o * (1 - o)
d1 = (d2 @ W2.T) * h * (1 - h)

W2 -= 0.1 * h.T @ d2
W1 -= 0.1 * X.T @ d1

print("Updated W1:\n", W1)
print("Updated W2:\n", W2)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

(x, y), (xt, yt) = tf.keras.datasets.mnist.load_data()

x = x / 255.0
xt = xt / 255.0

for use_bn in [False, True]:

    seq = [
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128)
    ]

    if use_bn:
        seq.append(layers.BatchNormalization())

    seq += [
        layers.Activation("relu"),
        layers.Dense(10, activation="softmax")
    ]

    model = models.Sequential(seq)

    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

    model.fit(x, y, epochs=3, verbose=0)

    print("Batch Normalization:", use_bn)
    print(model.evaluate(xt, yt, verbose=0))
```

```python
import numpy as np
from tensorflow.keras import layers, models

text = "deep learning makes computers learn patterns"

chars = sorted(set(text))
to_id = {c: i for i, c in enumerate(chars)}

seq = 20
X = []
Y = []

for i in range(len(text) - seq):
    X.append([to_id[c] for c in text[i:i+seq]])
    Y.append(to_id[text[i+seq]])

model = models.Sequential([
    layers.Embedding(len(chars), 16),
    layers.SimpleRNN(64),
    layers.Dense(len(chars), activation="softmax")
])

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy")

model.fit(np.array(X), np.array(Y), epochs=100, verbose=0)

seed = text[:seq]
result = seed

for i in range(30):
    ids = np.array([[to_id[c] for c in seed]])
    next_char = chars[np.argmax(model.predict(ids, verbose=0)[0])]
    result += next_char
    seed = seed[1:] + next_char

print(result)
```

```python
import numpy as np
from tensorflow.keras import layers, models

encoder_input = layers.Input(shape=(5, 8))

_, h, c = layers.LSTM(
    32, return_state=True
)(encoder_input)

decoder_input = layers.Input(shape=(5, 8))

decoder = layers.LSTM(
    32, return_sequences=True
)(decoder_input, initial_state=[h, c])

output = layers.Dense(8, activation="softmax")(decoder)

model = models.Model(
    [encoder_input, decoder_input], output
)

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy"
)

X = np.random.rand(100, 5, 8)
Y = np.random.rand(100, 5, 8)

model.fit([X, Y], Y, epochs=2, verbose=0)

print("Model Trained Successfully")
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X, y = load_breast_cancer(return_X_y=True)

X = StandardScaler().fit_transform(X)

X1, X2, y1, y2 = train_test_split(
    X, y, test_size=0.2, random_state=1
)

model1 = RandomForestClassifier(random_state=1)
model1.fit(X1, y1)

print("Before PCA:", accuracy_score(y2, model1.predict(X2)))

pca = PCA(n_components=0.95)

A = pca.fit_transform(X1)
B = pca.transform(X2)

model2 = RandomForestClassifier(random_state=1)
model2.fit(A, y1)

print("After PCA:", accuracy_score(y2, model2.predict(B)))
```

```python
import numpy as np
from sklearn.decomposition import FastICA

t = np.linspace(0, 8, 1000)

S = np.c_[
    np.sin(2*t),
    np.sign(np.sin(3*t))
]

A = np.array([[1, 2], [2, 1]])

mixed = S @ A.T

ica = FastICA(n_components=2, random_state=1)

separated = ica.fit_transform(mixed)

print("Mixed Shape:", mixed.shape)
print("Separated Shape:", separated.shape)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import FastICA

t = np.linspace(0, 8, 1000)

S = np.c_[
    np.sin(2*t),
    np.sign(np.sin(3*t))
]

A = np.array([[1, 2], [2, 1]])

X = S @ A.T

ica = FastICA(n_components=2, random_state=1)

sources = ica.fit_transform(X)

plt.plot(sources)
plt.title("ICA Separated Sources")
plt.show()

print("Reconstruction Error:",
      np.mean((X - ica.inverse_transform(sources))**2))
```

```python
import numpy as np
from sklearn.decomposition import PCA, FactorAnalysis
from sklearn.metrics import mean_squared_error

X = np.random.default_rng(1).normal(size=(200, 6))

pca = PCA(n_components=3)
Z1 = pca.fit_transform(X)
R1 = pca.inverse_transform(Z1)

fa = FactorAnalysis(n_components=3, random_state=1)
Z2 = fa.fit_transform(X)
R2 = Z2 @ fa.components_ + fa.mean_

print("PCA Error:", mean_squared_error(X, R1))
print("Factor Analysis Error:", mean_squared_error(X, R2))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

X, y = load_breast_cancer(return_X_y=True)

X = StandardScaler().fit_transform(X)

X1, X2, y1, y2 = train_test_split(
    X, y, test_size=0.2,
    random_state=1, stratify=y
)

model = models.Sequential([
    layers.Input(shape=(X.shape[1],)),
    layers.Dense(32, activation="relu"),
    layers.Dropout(0.2),
    layers.Dense(16, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.fit(X1, y1, epochs=20, batch_size=16,
          validation_split=0.2, verbose=0)

pred = (model.predict(X2).ravel() >= 0.5).astype(int)

print("Evaluation:", model.evaluate(X2, y2))
print("Confusion Matrix:\n", confusion_matrix(y2, pred))
print(classification_report(y2, pred))
```
