```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

X, y = make_classification(n_samples=500, n_features=20,
                           random_state=1)

X1, X2, y1, y2 = train_test_split(X, y, random_state=1)

for solver in ["sgd", "adam"]:
    model = MLPClassifier(
        hidden_layer_sizes=(32,),
        solver=solver,
        max_iter=500,
        random_state=1
    )

    model.fit(X1, y1)

    print(solver, "Accuracy:", accuracy_score(y2, model.predict(X2)))
    print("Iterations:", model.n_iter_)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

model = models.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.fit(x_train, y_train, epochs=3)

print("Test:", model.evaluate(x_test, y_test))
```

```python
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix

(x, y), (xt, yt) = tf.keras.datasets.mnist.load_data()

x = x[..., None] / 255.0
xt = xt[..., None] / 255.0

model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(16, 3, activation="relu",
                           input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(10, activation="softmax")
])

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.fit(x, y, epochs=2, verbose=0)

pred = model.predict(xt).argmax(axis=1)

print("Confusion Matrix:\n", confusion_matrix(yt, pred))
print(classification_report(yt, pred))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import urllib.request
import zipfile
import os

zip_path = 'plant_disease.zip'
url = 'https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz'
if not os.path.exists('plant_photos'):
    print("Downloading image dataset...")
    os.system('wget -q https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz')
    os.system('tar -xzf flower_photos.tgz')
    os.system('mv flower_photos plant_photos')

    os.makedirs('plant_disease/train', exist_ok=True)
    os.makedirs('plant_disease/test', exist_ok=True)
    import shutil
    for folder in os.listdir('plant_photos'):
        if os.path.isdir(os.path.join('plant_photos', folder)):
            shutil.copytree(os.path.join('plant_photos', folder), os.path.join('plant_disease/train', folder), dirs_exist_ok=True)
            shutil.copytree(os.path.join('plant_photos', folder), os.path.join('plant_disease/test', folder), dirs_exist_ok=True)
    print("Dataset prepared successfully.")

train = tf.keras.utils.image_dataset_from_directory(
    "plant_disease/train", image_size=(224, 224), batch_size=32
)

test = tf.keras.utils.image_dataset_from_directory(
    "plant_disease/test", image_size=(224, 224), batch_size=32
)

base = tf.keras.applications.ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base.trainable = False

inputs = layers.Input((224, 224, 3))
x = tf.keras.applications.resnet50.preprocess_input(inputs)
x = base(x, training=False)
x = layers.GlobalAveragePooling2D()(x)

outputs = layers.Dense(
    len(train.class_names), activation="softmax"
)(x)

model = models.Model(inputs, outputs)

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.fit(train, epochs=2, validation_data=test)
print(model.evaluate(test))
```

```python
import numpy as np
from tensorflow.keras import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

temp = np.sin(np.arange(600) * 0.05)

X = np.array([
    temp[i:i+10] for i in range(len(temp)-10)
])[..., None]

y = temp[10:]

model = Sequential([
    SimpleRNN(32, input_shape=(10, 1)),
    Dense(1)
])

model.compile(optimizer="adam", loss="mse")
model.fit(X, y, epochs=10, verbose=0)

print("Next Temperature:", model.predict(X[-1:]))
```
