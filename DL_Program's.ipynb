```python
import numpy as np

image = np.array([
    [150, 200],
    [50,  100]
])

input_vector = image.flatten()
print(f"1. Flattened Input Vector (shape {input_vector.shape}):\n{input_vector}\n")

np.random.seed(42)
weights = np.random.uniform(-0.5, 0.5, size=(3, 4))
biases = np.array([2.0, -1.0, 0.5])

print(f"Weights Matrix (shape {weights.shape}):\n{weights.round(2)}\n")
print(f"Biases (shape {biases.shape}):\n{biases}\n")


layer_output = np.dot(weights, input_vector) + biases

print(f"2. Matrix Multiplication result (Weights * Input):\n{(np.dot(weights, input_vector)).round(2)}\n")
print(f"3. Final Output after adding Biases (shape {layer_output.shape}):\n{layer_output.round(2)}")
```

Covariance, Eigenvalues and Eigenvectors Analysis

```python
import numpy as np
import matplotlib.pyplot as plt

study_hours = np.array([2, 4, 6, 8, 10])
marks = np.array([35, 45, 60, 75, 85])

covariance_matrix = np.cov(study_hours, marks, ddof=1)
sample_covariance = covariance_matrix[0, 1]

print(f"Study Hours: {study_hours}")
print(f"Examination Marks: {marks}")
print(f"Calculated Sample Covariance: {sample_covariance:.2f}\n")

plt.figure(figsize=(8, 5))
plt.scatter(study_hours, marks, color='blue', marker='o', s=100, label='Student Data')
plt.plot(study_hours, np.poly1d(np.polyfit(study_hours, marks, 1))(study_hours), color='red', linestyle='--', label='Trend Line')
plt.title('Relationship Between Study Hours and Exam Marks')
plt.xlabel('Study Hours')
plt.ylabel('Examination Marks')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
temperature = np.array([22, 25, 28, 30, 32])
humidity = np.array([80, 75, 68, 60, 55])

covariance_matrix = np.cov(temperature, humidity, ddof=1)

print(f"Temperature: {temperature}")
print(f"Humidity: {humidity}\n")
print(f"Covariance Matrix:\n{covariance_matrix}\n")
print(f"Covariance between Temperature and Humidity: {covariance_matrix[0, 1]:.2f}")
plt.figure(figsize=(8, 5))
plt.scatter(temperature, humidity, color='teal', marker='s', s=100, label='Recorded Data')
plt.plot(temperature, np.poly1d(np.polyfit(temperature, humidity, 1))(temperature), color='coral', linestyle='--', label='Trend Line')
plt.title('Relationship Between Temperature and Humidity')
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([
    [4, 2],
    [2, 4]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Matrix A:\n{A}\n")
print(f"Calculated Eigenvalues: {eigenvalues}")

plt.figure(figsize=(6, 4))
x_labels = [f'λ{i+1}\n({val:.1f})' for i, val in enumerate(eigenvalues)]
plt.bar(x_labels, eigenvalues, color=['skyblue', 'salmon'], edgecolor='black', width=0.4)
plt.title('Eigenvalues of Matrix A')
plt.ylabel('Magnitude')
plt.ylim(0, max(eigenvalues) + 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

C = np.array([
    [3, 1],
    [1, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(C)

print(f"Covariance Matrix C:\n{C}\n")
for i in range(len(eigenvalues)):
    print(f"Eigenvalue λ{i+1} = {eigenvalues[i]:.2f}")
    print(f"Eigenvector v{i+1} = {eigenvectors[:, i]}\n")

plt.figure(figsize=(6, 6))
plt.axhline(0, color='grey', linestyle='--', linewidth=1)
plt.axvline(0, color='grey', linestyle='--', linewidth=1)

colors = ['red', 'blue']
for i in range(len(eigenvalues)):

    dx, dy = eigenvectors[:, i] * eigenvalues[i]
    plt.quiver(0, 0, dx, dy, angles='xy', scale_units='xy', scale=1,
               color=colors[i], label=f'v{i+1} (λ={eigenvalues[i]:.1f})')

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.gca().set_aspect('equal', adjustable='box')
plt.title('Eigenvectors of Covariance Matrix C')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

C_sensor = np.array([
    [5, 2],
    [2, 2]
])

eigenvalues, eigenvectors = np.linalg.eig(C_sensor)


largest_idx = np.argmax(eigenvalues)
principal_val = eigenvalues[largest_idx]
principal_vec = eigenvectors[:, largest_idx]

print(f"Sensor Covariance Matrix C:\n{C_sensor}\n")
for i in range(len(eigenvalues)):
    is_principal = "(Principal Direction)" if i == largest_idx else ""
    print(f"Eigenvalue λ{i+1} = {eigenvalues[i]:.4f} {is_principal}")
    print(f"Eigenvector v{i+1} = {eigenvectors[:, i]}\n")

plt.figure(figsize=(7, 7))
plt.axhline(0, color='grey', linestyle='--', linewidth=1)
plt.axvline(0, color='grey', linestyle='--', linewidth=1)

colors = ['crimson' if i == largest_idx else 'royalblue' for i in range(len(eigenvalues))]
for i in range(len(eigenvalues)):
    dx, dy = eigenvectors[:, i] * eigenvalues[i]
    label_text = f'v{i+1} (λ={eigenvalues[i]:.2f})'
    if i == largest_idx:
        label_text += ' [Principal]'

    plt.quiver(0, 0, dx, dy, angles='xy', scale_units='xy', scale=1,
               color=colors[i], label=label_text, width=0.008)

max_lim = np.ceil(np.max(eigenvalues)) + 1
plt.xlim(-max_lim, max_lim)
plt.ylim(-max_lim, max_lim)
plt.gca().set_aspect('equal', adjustable='box')

plt.title('Eigenvectors and Principal Variation of Sensor Data')
plt.xlabel('Sensor 1 dimension')
plt.ylabel('Sensor 2 dimension')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

Probability Distributions and Bayes’ Theorem


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 20
p = 0.1

prob_3_defective = binom.pmf(3, n, p)
print(f"Probability of exactly 3 defective components: {prob_3_defective:.4f} ({prob_3_defective*100:.2f}%)\n")

x = np.arange(0, n + 1)
pmf_values = binom.pmf(x, n, p)

plt.figure(figsize=(10, 6))
plt.bar(x, pmf_values, color='skyblue', edgecolor='black', alpha=0.8, label='PMF')
plt.step(x, pmf_values, where='mid', color='red', linestyle='--', alpha=0.7, label='CDF Trend (Step)')
plt.title('Binomial Distribution (n=20, p=0.1) PMF')
plt.xlabel('Number of Defective Components')
plt.ylabel('Probability')
plt.xticks(x)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 70
std_dev = 10

prob_60 = norm.cdf(60, mean, std_dev)
prob_80 = norm.cdf(80, mean, std_dev)
interval_prob = prob_80 - prob_60

print(f"Probability of a student scoring between 60 and 80: {interval_prob:.4f} ({interval_prob*100:.2f}%)\n")

x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
y = norm.pdf(x, mean, std_dev)

plt.figure(figsize=(10, 6))
plt.plot(x, y, color='darkblue', linewidth=2, label='Normal Distribution (μ=70, σ=10)')

x_fill = np.linspace(60, 80, 500)
y_fill = norm.pdf(x_fill, mean, std_dev)
plt.fill_between(x_fill, y_fill, color='orange', alpha=0.5, label='60 to 80 (approx. 68.27%)')

plt.title('Normal Distribution of Exam Marks')
plt.xlabel('Examination Marks')
plt.ylabel('Probability Density')
plt.axvline(mean, color='red', linestyle='--', alpha=0.7, label=f'Mean (μ={mean})')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

mu = 4

prob_6_arrivals = poisson.pmf(6, mu)
print(f"Probability of exactly 6 customers arriving: {prob_6_arrivals:.4f} ({prob_6_arrivals*100:.2f}%)\n")

x = np.arange(0, 13)
pmf_values = poisson.pmf(x, mu)

plt.figure(figsize=(10, 6))
plt.bar(x, pmf_values, color='lightgreen', edgecolor='black', alpha=0.8, label='Poisson PMF (μ=4)')
plt.plot(x, pmf_values, color='darkgreen', marker='o', linestyle='-', alpha=0.7, label='PMF Line')
plt.title('Poisson Distribution of Customer Arrivals (Average μ = 4)')
plt.xlabel('Number of Customer Arrivals in 10-Minute Interval')
plt.ylabel('Probability')
plt.xticks(x)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt


prior_disease = 0.02
prior_no_disease = 0.98

sensitivity = 0.95
false_positive_rate = 0.05


p_positive = (sensitivity * prior_disease) + (false_positive_rate * prior_no_disease)


posterior_disease = (sensitivity * prior_disease) / p_positive

print(f"Prior Probability of having the disease: {prior_disease*100:.2f}%")
print(f"Total Probability of testing positive: {p_positive*100:.2f}%")
print(f"Posterior Probability of actually having the disease given a positive test: {posterior_disease*100:.2f}%\n")


labels = ['Prior Probability P(D)', 'Posterior Probability P(D | Pos)']
values = [prior_disease, posterior_disease]
colors = ['coral', 'royalblue']

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, values, color=colors, edgecolor='black', width=0.5)

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.01,
             f'{height*100:.2f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.title("Comparison of Prior vs. Posterior Probability (Bayes' Theorem)", fontsize=14, pad=15)
plt.ylabel('Probability', fontsize=12)
plt.ylim(0, max(values) + 0.1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
prior_spam = 0.30
prior_ham = 0.70
p_prize_given_spam = 0.70
p_prize_given_ham = 0.10
p_prize = (p_prize_given_spam * prior_spam) + (p_prize_given_ham * prior_ham)
posterior_spam = (p_prize_given_spam * prior_spam) / p_prize
print(f"Prior Probability of Spam: {prior_spam*100:.2f}%")
print(f"Total Probability of containing 'prize': {p_prize*100:.2f}%")
print(f"Posterior Probability of Spam (given 'prize'): {posterior_spam*100:.2f}%\n")
labels = ['Prior Probability P(Spam)', 'Posterior Probability P(Spam | Prize)']
values = [prior_spam, posterior_spam]
colors = ['lightcoral', 'indianred']
plt.figure(figsize=(8, 5))
bars = plt.bar(labels, values, color=colors, edgecolor='black', width=0.5)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.01,
             f'{height*100:.2f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')
plt.title("Spam Filter: Prior vs. Posterior Probability (Bayes' Theorem)", fontsize=14, pad=15)
plt.ylabel('Probability', fontsize=12)
plt.ylim(0, max(values) + 0.1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
