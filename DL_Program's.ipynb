```python
import numpy as np

week1_sales = np.array([120, 85, 150, 90, 200, 110, 95])
week2_sales = np.array([130, 80, 160, 85, 195, 120, 100])
products = np.array(["Product A", "Product B", "Product C", "Product D", "Product E", "Product F", "Product G"])

total_sales_per_product = week1_sales + week2_sales

sales_change = week2_sales - week1_sales

fell_mask = sales_change < 0
falling_products = products[fell_mask]

print(f"Products: {list(products)}")
print(f"Week 1 Sales: {week1_sales}")
print(f"Week 2 Sales: {week2_sales}")
print(f"Total Sales for 2 Weeks: {total_sales_per_product}")
print(f"Sales Change (W2 - W1): {sales_change}")
print(f"Products with falling sales: {list(falling_products)}")
```

```python
import numpy as np

usd_prices = np.array([49.99, 99.99, 149.99, 29.99, 199.99])
exchange_rate = 83.5

inr_prices = usd_prices * exchange_rate

discounted_inr = inr_prices * 0.9

expected_inr = 0.9 * exchange_rate * usd_prices
verification_check = np.allclose(discounted_inr, expected_inr)

print(f"USD Prices: {usd_prices}")
print(f"INR Prices (Converted): {inr_prices}")
print(f"Discounted INR Prices (10% off): {discounted_inr}")
print(f"Verification Successful: {verification_check}")
```

```python
import numpy as np

image = np.array([
    [150, 200],
    [50,  100]
])

input_vector = image.flatten()
print(f"1. Flattened Input Vector (shape {input_vector.shape}):\n{input_vector}\n")

np.random.seed(42)
weights = np.random.uniform(-0.5, 0.5, size=(3, 4))
biases = np.array([2.0, -1.0, 0.5])

print(f"Weights Matrix (shape {weights.shape}):\n{weights.round(2)}\n")
print(f"Biases (shape {biases.shape}):\n{biases}\n")


layer_output = np.dot(weights, input_vector) + biases

print(f"2. Matrix Multiplication result (Weights * Input):\n{(np.dot(weights, input_vector)).round(2)}\n")
print(f"3. Final Output after adding Biases (shape {layer_output.shape}):\n{layer_output.round(2)}")
```

```python
import numpy as np

np.random.seed(42)


attendance = np.random.uniform(75, 100, size=(100, 1))
assignments = np.random.uniform(50, 100, size=(100, 1))
internals = np.random.uniform(10, 30, size=(100, 1))

X = np.hstack((attendance, assignments, internals))
print(f"Feature Matrix X Shape: {X.shape} (100 Students, 3 Features)")

W = np.array([[
    0.4,
    0.4,
    0.2
]]).T

bias = 5.0

predictions = np.dot(X, W) + bias

print("\nFirst 5 Students (Attendance, Assignments, Internals):")
print(X[:5].round(2))
print("\nFirst 5 Predicted Performance Scores:")
print(predictions[:5].round(2))
```

```python
import numpy as np

X = np.array([4, 6, 8])
W = np.array([0.5, 0.3, 0.2])
b = 2

predicted_output = np.dot(X, W) + b

print(f"Input Features X: {X}")
print(f"Weights W: {W}")
print(f"Bias b: {b}")
print(f"Predicted Output (W . X + b): {predicted_output}")
```

```python
import numpy as np

X = np.array([
    [30, 60],
    [35, 70]
])

W = np.array([
    [0.2, 0.5],
    [0.4, 0.3]
])

predicted_output = np.dot(X, W)

print(f"Input Matrix X:\n{X}\n")
print(f"Weight Matrix W:\n{W}\n")
print(f"Predicted Output Matrix (X . W):\n{predicted_output}")
```

```python
import numpy as np

W = np.array([0.7, 0.5, 0.9])
G = np.array([0.2, 0.1, 0.3])
learning_rate = 0.1

updated_W = W - learning_rate * G

print(f"Current Weights W: {W}")
print(f"Gradient G: {G}")
print(f"Learning Rate: {learning_rate}")
print(f"Updated Weights: {updated_W.round(4)}")
```

```python
import numpy as np

A = np.array([2, 4, 6])
B = np.array([1, 3, 5])

combined_representation = A + B

similarity_score = np.dot(A, B)

print(f"Movie A Vector: {A}")
print(f"Movie B Vector: {B}")
print(f"Combined Vector (Addition A + B): {combined_representation}")
print(f"Similarity Score (Dot Product A . B): {similarity_score}")
```
