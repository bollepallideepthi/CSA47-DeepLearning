```python
pA = 0.6
pB = 0.4
pDA = 0.02
pDB = 0.05

result = (pDA * pA) / ((pDA * pA) + (pDB * pB))

print("Conditional Probability:", result)
```

```python
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90

positive = sensitivity * prevalence + (1 - specificity) * (1 - prevalence)

result = (sensitivity * prevalence) / positive

print("Bayes Probability:", result)
```

```python
import math

labels = ["Low", "Low", "High", "Low", "High", "High"]

p = [labels.count(x) / len(labels) for x in set(labels)]

entropy = -sum(x * math.log2(x) for x in p)

print("Entropy:", entropy)
```

```python
import numpy as np

actual = np.array([1, 0, 1, 1])
predicted = np.array([0.9, 0.2, 0.7, 0.6])

loss = -np.mean(
    actual * np.log(predicted) +
    (1-actual) * np.log(1-predicted)
)

print("Cross Entropy Loss:", loss)
```

```python
import numpy as np

P = np.array([0.4, 0.4, 0.2])
Q = np.array([0.5, 0.3, 0.2])

kl = np.sum(P * np.log(P / Q))

print("KL Divergence:", kl)
```

```python
x = 10
lr = 0.1

for i in range(100):
    gradient = 2 * (x - 3)
    x = x - lr * gradient

print("Minimum:", x)
print("Loss:", (x - 3) ** 2)
```

```python
import numpy as np

X = np.array([1., 2., 3., 4.])
y = 2 * X

def train(mode):
    w = 0.0

    for epoch in range(30):
        ids = np.arange(len(X))

        if mode == "SGD":
            np.random.shuffle(ids)
            batches = [[i] for i in ids]
        elif mode == "Mini-batch":
            np.random.shuffle(ids)
            batches = [ids[i:i+2] for i in range(0, 4, 2)]
        else:
            batches = [ids]

        for batch in batches:
            grad = np.mean(2 * X[batch] * (w * X[batch] - y[batch]))
            w -= 0.01 * grad

    return w

for mode in ["Batch", "SGD", "Mini-batch"]:
    print(mode, train(mode))
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1000, 2], [1200, 2], [1500, 3],
              [1800, 3], [2000, 4]])

y = np.array([30, 35, 45, 55, 60])

model = LinearRegression()
model.fit(X, y)

print("Prediction:", model.predict([[1600, 3]]))
print("Score:", model.score(X, y))
```
